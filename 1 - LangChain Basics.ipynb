### Example Selector

##### Input -> example selector -> few shot example

###### example selector -> 3 types, 1 - SemanticSimilarityExampleSelector, 2 - LengthBasedExampleSelector, 3 - MMR

```python
from langchain_core.example_selectors import SemanticSimilarityExampleSelector,MaxMarginalRelevanceExampleSelector
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
examples = [
    {
        "input": "How can I reverse a list in Python?",
        "output": "Use list.reverse() or reversed()."
    },
    {
        "input": "How can I sort a list in Python?",
        "output": "Use list.sort() or sorted()."
    },
    {
        "input": "How can I create a dictionary in Python?",
        "output": "Use curly braces {}."
    },
    {
        "input": "How can I read a CSV file using pandas?",
        "output": "Use pandas.read_csv()."
    },
    {
        "input": "How can I remove duplicates from a DataFrame?",
        "output": "Use DataFrame.drop_duplicates()."
    },
]
```

```python
example_prompt = PromptTemplate.from_template(
    "Input: {input}\n"
    "Output: {output}"
)
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)

selector = SemanticSimilarityExampleSelector.from_examples(
    examples=examples,
    embeddings=embeddings,
    vectorstore_cls=FAISS,
    k=3,
)
```

```python
prompt = FewShotPromptTemplate(
    example_selector=selector,
    example_prompt=example_prompt,
    prefix="Classify the sentiment of the following text:",
    suffix="Input: {input}\nOutput:",
    input_variables=["input"],
)


# Test
result = prompt.invoke({
    "input": "This movie was excellent"
})

print(result.to_string())
```

```python
# Examples
examples = [
    {
        "input": "I love this movie",
        "output": "Positive"
    },
    {
        "input": "This movie was amazing",
        "output": "Positive"
    },
    {
        "input": "The film was fantastic",
        "output": "Positive"
    },
    {
        "input": "I enjoyed watching this film",
        "output": "Positive"
    },
    {
        "input": "The movie was terrible",
        "output": "Negative"
    },
    {
        "input": "I hated this movie",
        "output": "Negative"
    },
]


example_prompt = PromptTemplate.from_template(
    "Input: {input}\nOutput: {output}"
)


embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)


example_selector = MaxMarginalRelevanceExampleSelector.from_examples(
    examples,
    embeddings,
    FAISS,
    k=3,
)


prompt = FewShotPromptTemplate(
    example_selector=example_selector,
    example_prompt=example_prompt,
    prefix="Classify the sentiment of the following text:",
    suffix="Input: {input}\nOutput:",
    input_variables=["input"],
)


# Test
result = prompt.invoke({
    "input": "This movie was excellent"
})

print(result.to_string())
```
