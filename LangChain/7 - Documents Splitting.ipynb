## Character Text Splitter

```python
from langchain_text_splitters import CharacterTextSplitter

# Chunk Size: Character Count
# Chunk Overlap: (a b c d e f) (e f g h i j)
c_splitter = CharacterTextSplitter(
    chunk_size=26,
    chunk_overlap=4
)
```

```python
# The Default Separator Is "\n"
# We Don't Have Any "\n" Here So It Will Do Nothing
text = "a b c d e f g h i j k l m n o p q r s t u v w x y z"
chunks = c_splitter.split_text(text)
print(chunks)
```

```python
c_splitter = CharacterTextSplitter(
    separator=" ",
    chunk_size=26,
    chunk_overlap=4
)
chunks = c_splitter.split_text(text)
print(chunks)
```

## Compare 

```python
text = """
Artificial Intelligence is transforming healthcare by helping doctors
detect diseases earlier and analyze medical images more efficiently.

Machine learning models can analyze large amounts of medical data
and help researchers discover new patterns.
"""
```

```python
c_splitter = CharacterTextSplitter(
    separator=" ",
    chunk_size=100,
    chunk_overlap=0
)

chunks = c_splitter.split_text(text)

for i, chunk in enumerate(chunks, start=1):
    print(f"\nChunk {i}:")
    print(chunk)
```

```python
r_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=0,
    separators=["\n\n", "\n", " ", ""]
)

chunks = r_splitter.split_text(text)

for i, chunk in enumerate(chunks, start=1):
    print(f"\nChunk {i}:")
    print(chunk)
```

## Token Text Splitter

```python
from langchain_text_splitters import TokenTextSplitter

# In This Splitter (chunk_size = Tokens)

text_splitter = TokenTextSplitter(
    chunk_size=100,
    chunk_overlap=10
)
```

```python
text = "Artificial intelligence is changing the world."

chunks = text_splitter.split_text(text)

print(chunks)
```

## Markdown Header Text Splitter

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter
```

```python
markdown_document = """
# AI Course

## Introduction

This course teaches Artificial Intelligence basics.

## Machine Learning

This section teaches machine learning algorithms.

### Supervised Learning

This section explains classification and regression.

## Deep Learning

This section explains neural networks.
"""
```

```python
headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
    ("###", "Header 3"),
]
```

```python
markdown_splitter = MarkdownHeaderTextSplitter(headers_to_split_on = headers_to_split_on)
chunks = markdown_splitter.split_text(markdown_document)
```

```python
for chunk in chunks:
    print("\nCONTENT:")
    print(chunk.page_content)

    print("METADATA:")
    print(chunk.metadata)
```
