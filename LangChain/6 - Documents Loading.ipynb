## PDF Loader

```python
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
pdf_path = r"assets\Bishoy_Amgad_CV.pdf"
pdf_loader = PyMuPDFLoader(pdf_path)
```

```python
pdf_content = pdf_loader.load()
```

```python
print(type(pdf_content))
print(len(pdf_content))
```

```python
print(pdf_content[1].page_content)
```

## Excel Loader

## Youtube Transcript

```python
from langchain_community.document_loaders import YoutubeLoader
```

```python
url = "https://youtu.be/Vv6e2Rb1Q6w?si=7aoshYpQkFsLijTA"
yt_loader = YoutubeLoader.from_youtube_url(url, language = ["ar"])
```
