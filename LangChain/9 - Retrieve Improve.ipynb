## How To Get Better Retrieve ?

### 1 - Maximum Marginal Relevance (MMR)

#### It Use When We Need To Get Variety In Data

#### What If All Retrieved Chunks All Of Them Is Positive -> The Model Don't See Any Negative

#### What If All Retrieved Chunks Have The Same Information Nothing New

results = vector_db.max_marginal_relevance_search(
    query,
    k=4,
    fetch_k=10
)

fetch_k: Get Top 10

k: Retrieve Only K

### 2 - Filters

#### Instead To Get The Model Search In All Chunks

#### We Can Tell Him Where To Search

results = vector_db.similarity_search(
    query,
    k=5,
    filter={"lecture": 3}
)

### 3 - Compression

#### It Use Extra LLM To Compress The Retrieval Chunks

#### Decrease The Context But Increase The Cost

### 4 - Combining Various Retrieval Techniques

#### Combine Many Retriever At Same Time

#### Ex: MMR + Filter + Compress
